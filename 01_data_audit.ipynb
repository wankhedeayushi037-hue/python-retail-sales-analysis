**Approach:**


# Phase 2: Data Cleaning and Preparation

#### Step 1: Create a Working Copy and Handle Missing Descriptions

### Objective
Create a separate working DataFrame and recover missing product descriptions wherever a matching StockCode has an available description.

#### Approach
1. Copy the original DataFrame into df_clean.
2. Build a mapping between StockCode and an available Description.
3. Fill missing descriptions using the mapping.
4. Count the descriptions that remain missing.

#### Why It Matters
This preserves the original dataset and improves product information for subsequent analysis.

#### Data Quality Note
The mapping selects the first available description for each StockCode. This assumes that the selected description is suitable for other rows with the same code. Some descriptions may remain missing if no matching description is available.


```python
# Create a working copy of the original dataset
df_clean = df.copy()


# Build a StockCode-to-Description mapping
description_map = (
    df_clean.dropna(subset=["Description"])
    .drop_duplicates(subset=["StockCode"])
    .set_index("StockCode")["Description"]
)

# Fill missing descriptions using matching StockCodes
df_clean["Description"] = df_clean["Description"].fillna(
    df_clean["StockCode"].map(description_map)
)

# Check remaining missing descriptions
print(
    "Missing descriptions after filling:",
    df_clean["Description"].isna().sum()
)
```


#### Step 2: Remove Exact Duplicate Records

#### Objective
Remove exact duplicate rows from the working dataset to reduce the risk of counting identical records more than once.

#### Approach
Use Pandas drop_duplicates() to remove rows that are identical across all columns, then check the remaining row count.

#### Why It Matters
Duplicate transactions can distort sales totals and other business metrics.

#### Validation
Compare the row count after cleaning with the original dataset and confirm how many duplicate rows were removed.


```python
# Remove exact duplicate rows
df_clean = df_clean.drop_duplicates()

print("Rows after removing duplicates:", len(df_clean))
```


#### Step 3: Calculate Transaction Value

#### Objective
Calculate the transaction value for each row using quantity and unit price.

#### Approach
Multiply Quantity by Price and store the result in a new column named TransactionValue.

#### Formula
TransactionValue = Quantity × Price

#### Why It Matters
This column will be used to calculate sales totals, compare transaction values, and analyze business performance.

#### Validation
Display a sample of Quantity, Price, and TransactionValue to verify that the calculation is applied correctly.

##### Data Quality Note
TransactionValue may be negative when Quantity is negative. This value represents the signed transaction amount, not profit.


```python
# Calculate transaction value for each row
df_clean["TransactionValue"] = (
    df_clean["Quantity"] * df_clean["Price"]
)

# Display sample results
print(
    df_clean[["Quantity", "Price", "TransactionValue"]].head()
)
```


#### Step 4: Validate Total Quantity and Net Transaction Value

#### Objective
Calculate the net quantity and net transaction value in the cleaned dataset.

#### Approach
- Sum the Quantity column across all remaining records.
- Sum TransactionValue, including both positive and negative values.

#### Why It Matters
These totals provide a baseline for checking the cleaned dataset and understanding the effect of negative transactions.

#### Interpretation Note
Net quantity is not necessarily the number of units sold to customers because negative quantities may represent returns or other adjustments. Net transaction value is not the same as confirmed sales revenue or profit.


```python
# Calculate the net quantity across all records, including returns
print("Total quantity sold:", df_clean["Quantity"].sum())

# Calculate the net transaction value after accounting for negative values
print(
    "Net transaction value:",
    round(df_clean["TransactionValue"].sum(), 2)
)
```


#### Step 5: Calculate Transaction Value for Positive-Quantity Records

#### Objective
Create a Boolean column named IsCancellation to identify invoices whose invoice number starts with "C".

#### Approach
- Mark matching invoices as True.
- Mark all other invoices as False.
- Count the records in both groups.

#### Why It Matters
A cancellation indicator makes it easier to filter and compare cancellation transactions with regular transactions during analysis.

#### Validation
Check the number of True and False values to confirm that the indicator has been created across the cleaned dataset.

#### Interpretation Note
This flag is based on the invoice prefix. It is a classification rule for this project and should not be treated as independent proof of every transaction's business meaning.


#### Step 10: Prepare the Sales Analysis Dataset

#### Objective
Create a separate dataset containing eligible sales transactions for business analysis.

#### Filtering Rules
- Exclude invoices flagged as cancellations.
- Keep records with Quantity greater than zero.
- Keep records with Price greater than zero.

#### Why It Matters
These filters help isolate regular sales transactions from cancellation records and transactions with non-positive quantities or prices.

#### Approach
Apply the filtering conditions to df_clean and create an independent copy named df_sales.

#### Validation
Check the number of retained rows and calculate their total TransactionValue.

#### Interpretation Note
The resulting total represents sales value under the project's filtering rules. It is not profit, and the filters should be documented when presenting results.


```python
# Keep only normal sales, excluding cancellation invoices
# Also exclude rows with zero or negative quantity or price
df_sales = df_clean[
    (~df_clean["IsCancellation"]) &
    (df_clean["Quantity"] > 0) &
    (df_clean["Price"] > 0)
].copy()  # Create a separate DataFrame without modifying df_clean

# Display the number of valid sales rows
print("Sales rows:", len(df_sales))

# Calculate and display the total sales value
print("Sales value:", round(df_sales["TransactionValue"].sum(), 2))
```

#### 3.1 Monthly Sales Revenue Analysis

**Objective:** Analyze monthly sales revenue to identify high-performing and low-performing months.

**Approach:**
- Extract the year and month from `InvoiceDate`.
- Group sales transactions by month.
- Calculate the total revenue using `TransactionValue`.

**Business Questions:**
- Which month generated the highest revenue?
- Which month generated the lowest revenue?
- How did sales revenue change over time?

**Business Value:** Helps identify sales trends and potential seasonality to support business planning and sales forecasting.

**Interpretation Note:** December 2010 has comparatively low revenue. Check whether the dataset covers the full month before interpreting this as a decline in sales.


```python
# Extract year and month from InvoiceDate and store it in a new column
df_sales["Month"] = df_sales["InvoiceDate"].dt.to_period("M")

# Group sales records month-wise and select the revenue column
monthly_sales = df_sales.groupby("Month")["TransactionValue"].sum()  # Calculate total revenue for each month

# Display monthly sales revenue
monthly_sales
```

#### 3.2 Identify Highest and Lowest Revenue Months

**Objective:** Identify the months with the highest and lowest total sales revenue.

**Approach:**
- Use `idxmax()` to identify the month with the highest revenue.
- Use `max()` to retrieve its revenue value.
- Use `idxmin()` and `min()` to identify the month with the lowest revenue and its value.

**Business Questions:**
- When did the business achieve its highest monthly revenue?
- Which month recorded the lowest revenue?

**Business Value:** Helps identify peak sales periods and months that may require further investigation.

**Interpretation Note:** A low-revenue month does not necessarily indicate poor performance; consider dataset coverage and seasonal patterns before drawing conclusions.


```python
# Find the month with the highest sales revenue and its value
print("Highest month:", monthly_sales.idxmax(), round(monthly_sales.max(), 2))

# Find the month with the lowest sales revenue and its value
print("Lowest month:", monthly_sales.idxmin(), round(monthly_sales.min(), 2))
```


#### 3.3 Top 10 Products by Revenue

**Objective:** Identify the ten products generating the highest total sales revenue.

**Approach:**
- Group sales transactions by product description.
- Calculate total `TransactionValue` for each product.
- Sort products in descending order of revenue.
- Select the top 10 products.

**Business Questions:**
- Which products generate the most revenue?
- Which products contribute most to the business's sales?

**Business Value:** Helps prioritize high-revenue products for inventory planning, product promotion, and sales strategy.

**Interpretation Note:** Revenue ranking does not indicate profitability. Product costs and margins would be needed to determine which products are most profitable.


```python
# Group sales records product-wise and select the revenue column
top_products = (
    df_sales.groupby("Description")["TransactionValue"]
    .sum()  # Calculate total revenue for each product
    .sort_values(ascending=False)  # Sort products by revenue, highest first
    .head(10)  # Select the top 10 products
)

# Display the top 10 products and their revenue
top_products
```


#### 3.4 Refined Top 10 Products by Revenue

**Objective:** Identify the top ten revenue-generating products after excluding transactions labelled `Manual`.

**Approach:**
- Exclude rows where `Description` is `Manual`.
- Group the remaining sales records by product description.
- Calculate total revenue using `TransactionValue`.
- Sort revenue in descending order and select the top ten products.

**Business Questions:**
- Which products generate the highest revenue after excluding `Manual` entries?
- Which products should receive greater attention in sales and inventory planning?

**Business Value:** Provides a more focused view of revenue-generating items by removing `Manual` entries from the product ranking.

```python
# Exclude Manual entries and group sales records product-wise
# Select TransactionValue to calculate revenue for each product
top_products = (
    df_sales[df_sales["Description"] != "Manual"]
    .groupby("Description")["TransactionValue"]
    .sum()  # Calculate total revenue generated by each product
    .sort_values(ascending=False)  # Sort revenue from highest to lowest
    .head(10)  # Select the top 10 revenue-generating products
)

# Display the top 10 products and their revenue
top_products
```


#### 3.8 Average Order Value (AOV) Analysis

**Objective:** Calculate the average revenue generated per invoice to understand the typical order value.

**Approach:**
- Group sales transactions by `Invoice`.
- Calculate the total transaction value for each invoice.
- Calculate the mean of invoice-level revenue to obtain AOV.

**Business Questions:**
- What is the average revenue generated per order?
- Can the business increase order value through product bundling or cross-selling?

**Business Value:** AOV helps evaluate purchasing behavior and identify opportunities to increase revenue per order.

**Interpretation Note:** AOV is calculated here as the arithmetic mean of invoice-level revenue. It represents average order revenue, not profit, and can be influenced by unusually large orders.


```python
# Calculate total revenue for each invoice
invoice_sales = df_sales.groupby("Invoice")["TransactionValue"].sum()

# Calculate the average revenue per invoice
average_order_value = invoice_sales.mean()

# Display the Average Order Value (AOV)
print("Average Order Value:", round(average_order_value, 2))
```

**Interpretation Note:** These counts come from two separately defined datasets. 
#### 3.11 Cancellation Invoice Rate Analysis

**Objective:** Calculate the percentage of cancellation invoices relative to the combined count of cancellation and sales invoices.

**Approach:**
- Add the unique cancellation invoice count and sales invoice count.
- Divide cancellation invoices by the combined invoice count.
- Multiply by 100 to express the result as a percentage.

**Business Question:** What proportion of the combined invoice count consists of cancellation invoices?

**Business Value:** Provides an initial indicator of cancellation activity and can help identify the need for further investigation.

**Interpretation Note:** This is a cancellation invoice share based on two separately counted invoice groups, not necessarily a true order cancellation rate. The same order may have a cancellation invoice and a sales invoice, and the two counts may overlap conceptually. Interpret the result cautiously.


```python
# Calculate total unique invoices, including cancellations
total_invoices = cancelled_invoices + normal_invoices

# Calculate the percentage of invoices that were cancelled
cancellation_rate = (cancelled_invoices / total_invoices) * 100

# Display the cancellation rate
print("Cancellation rate (%):", round(cancellation_rate, 2))
```
